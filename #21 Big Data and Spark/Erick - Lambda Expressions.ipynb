# Lambda Expressions review

```python
def cuadrado (num):
    result = num**2
    return result
```

```python
cuadrado(4)
```

```python
def cuadrado(num):
    return num**2
```

```python
cuadrado(4)
```

esto lo puedo hacer en una sola linea y funciona, pero ahora veremos como llevarlo a lambda expresion

```python
lambda num: num**2
```

```python
sq = lambda num: num**2
```

```python
sq(5)
```

```python
# si queremos evaluar si el resto de un numero es 0
even = lambda num: num%2==0
even(3)
```

```python
first = lambda texto: texto[0]
first("Hola")
```

```python
reversed_str = lambda texto: texto[::-1]
```

```python
reversed_str("Hola")
```

```python
def adder (x,y):
    return x+y

adder(5,4)
```

```python
adder = lambda x,y: x+y
adder(4,3)
```
