```python
# --- Definir tensores ---
x = tf.constant([10, 20, 30], dtype=tf.float32)
y = tf.constant([1, 2, 3], dtype=tf.float32)

# --- Operaciones ---
add = x + y            # Suma
sub = x - y            # Resta
mul = x * y            # Multiplicación
div = x / y            # División

# --- Imprimir resultados ---
print("Operations with Constants")
print("Addition:", add.numpy())
print("Subtraction:", sub.numpy())
print("Multiplication:", mul.numpy())
print("Division:", div.numpy())
```
