### Desde aqui va el codigo actualizado que se hace para hacer lo mismo pero con tensor flow 2.X

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import tensorflow as tf
```

```python

# ============================
# 1. Cargar datos
# ============================
df = pd.read_csv("iris.csv")
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'target']

X = df.drop("target", axis=1)
y = df["target"].astype(int)   # asegurar enteros

# Train/Test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python

# ============================
# 2. Definir modelo en Keras
# ============================
model = tf.keras.Sequential([
    tf.keras.layers.Dense(5, activation="relu", input_shape=(X.shape[1],)),
    tf.keras.layers.Dense(10, activation="relu"),
    tf.keras.layers.Dense(5, activation="relu"),
    tf.keras.layers.Dense(3, activation="softmax")  # 3 clases
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",  # porque y son enteros
    metrics=["accuracy"]
)
```

tf.keras.Sequential([...])

Crea un modelo secuencial, es decir, una pila de capas donde la salida de una es la entrada de la siguiente.

Es lo más simple de usar cuando el modelo es lineal (sin ramas ni múltiples entradas/salidas).

🔹 tf.keras.layers.Dense(10, activation="relu", input_shape=(X.shape[1],))

Capa densa (fully connected / perceptrón): cada neurona se conecta con todas las entradas.

10 → número de neuronas en esta capa.

activation="relu" → la función de activación es ReLU (Rectified Linear Unit), muy usada porque ayuda al entrenamiento profundo.

input_shape=(X.shape[1],) → forma de los datos de entrada.

X.shape[1] = número de features (en iris son 4: sepal_length, sepal_width, petal_length, petal_width).

Esto le dice a la red cuántas entradas tiene cada muestra.

```python

# ============================
# 3. Entrenamiento
# ============================
history = model.fit(
    X_train, y_train,
    validation_split=0.2,
    epochs=50,
    batch_size=10,
    verbose=0
)
```

✅ model → contiene el modelo entrenado con sus pesos ajustados.

✅ history → contiene el registro de métricas y pérdidas durante el entrenamiento (sirve para análisis, gráficas de aprendizaje, etc.).

Por defecto, cuando entrenas con arrays (X_train, y_train), Keras ya hace shuffle automático en cada epoch.

Es como si internamente hiciera:

dataset = dataset.shuffle(buffer_size=len(X_train))


Entonces no necesitas poner shuffle=True explícitamente aquí, porque ya está incluido.

```python

# ============================
# 4. Predicciones
# ============================
y_pred = model.predict(X_test)
final_preds = y_pred.argmax(axis=1)
```

```python

# ============================
# 5. Evaluación
# ============================
print(confusion_matrix(y_test, final_preds))
print(classification_report(y_test, final_preds))
```
